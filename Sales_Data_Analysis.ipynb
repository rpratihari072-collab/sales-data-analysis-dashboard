```python
import pandas as pd

# Load the train.csv file (make sure you uploaded it to Colab first)
df = pd.read_csv('train.csv')

# Look at the column names and data types
df.info()
```

```python
# 1. Convert Date columns to actual Date objects
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)

# 2. Fill missing Postal Codes with a placeholder (Power BI doesn't like empty values in maps)
df['Postal Code'] = df['Postal Code'].fillna(0).astype(int)

# 3. Remove any completely duplicate rows
df.drop_duplicates(inplace=True)

# 4. Create a 'Profit' column if it's missing (Optional, based on your version of the file)
# If the file only has 'Sales', you can simulate profit as 15% of sales for the project:
if 'Profit' not in df.columns:
    df['Profit'] = df['Sales'] * 0.15

print("Data cleaning complete!")
```

```python
from google.colab import files

# Save the cleaned file
df.to_csv('cleaned_superstore.csv', index=False)

# Download it to your computer
files.download('cleaned_superstore.csv')
```
